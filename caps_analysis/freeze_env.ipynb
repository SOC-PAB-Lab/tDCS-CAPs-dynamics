```python
import sys
import importlib.metadata

LIBRARIES = {
    "numpy":      "numpy",
    "scipy":      "scipy",
    "pandas":     "pandas",
    "matplotlib": "matplotlib",
    "seaborn":    "seaborn",
    "nibabel":    "nibabel",
    "nilearn":    "nilearn",
    "sklearn":    "scikit-learn",
    "brainspace": "brainspace",
    "brainstat":  "brainstat",
    "joypy":      "joypy",
    "tqdm":       "tqdm",
    "termcolor":  "termcolor",
    "dotenv":     "python-dotenv",
}

python_version = f"{sys.version_info.major}.{sys.version_info.minor}.{sys.version_info.micro}"
print(f"Python: {python_version}")
print()

resolved = {}
missing = []

for import_name, pip_name in LIBRARIES.items():
    try:
        version = importlib.metadata.version(pip_name)
        resolved[pip_name] = version
        print(f"  {pip_name}=={version}")
    except importlib.metadata.PackageNotFoundError:
        missing.append(pip_name)
        print(f"  {pip_name}: NOT FOUND")

output_path = "environment.yml"

with open(output_path, "w") as f:
    f.write(f"name: tdcs-caps\n")
    f.write("channels:\n")
    f.write("  - conda-forge\n")
    f.write("  - defaults\n")
    f.write("dependencies:\n")
    f.write(f"  - python={python_version}\n")
    f.write("  - pip\n")
    f.write("  - pip:\n")
    for pip_name, version in resolved.items():
        f.write(f"    - {pip_name}=={version}\n")

print(f"\nSaved -> {output_path}")
if missing:
    print(f"Warning: not found: {missing}")
```
